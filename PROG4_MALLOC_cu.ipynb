```python
""" Aloocate 3 integers on the host(CPU) using either stack (int a=5, b =15,c;) or heap(malloc function)
    Now allocate 3 integes on device(CPU) using cudaMalloc() function
    Copy Values of a and b from host to device (GPU) using cudaMalloc() function
    Call the device kernel.
    Copy the result (c) back from device to host using cudaMemCpy()function.
    Display the result on CPU.

    1]Malloc - cudaMalloc - allocates the memory
    2] cudaMemcpy - copies the memory
    3] cudaFree - frees the memory
    4] cudaMemset - Sets memory to value.
"""
```

```python
%%writefile add.cu

#include "cuda_runtime.h"
#include "device_launch_parameters.h"
#include <stdio.h>

//cuda kernel to add two integers
__global__ void add(int *a, int *b, int *result)
{
    *result= *a + *b;
}

int main()
{

int host_b = 7;
int host_a = 5;
int host_result;

//variables device
int * device_a,* device_b,* device_result;

//allocate memory on device
cudaMalloc((void **)&device_a, sizeof(int));
cudaMalloc((void **)&device_b, sizeof(int));
cudaMalloc((void **)&device_result, sizeof(int));

//copy values from host to device
cudaMemcpy(device_a, &host_a, sizeof(int), cudaMemcpyHostToDevice);
cudaMemcpy(device_b, &host_b, sizeof(int), cudaMemcpyHostToDevice);

//Launch kernel
add<<<1,1>>>(device_a, device_b, device_result);

// Copy of the result from device to host
cudaMemcpy(&host_result, device_result, sizeof(int), cudaMemcpyDeviceToHost);

// Display the result
printf("Result: %d\n", host_result);


//free memory
cudaFree(device_a);
cudaFree(device_b);
cudaFree(device_result);

//copy result from device to host
cudaMemcpy(&host_result, device_result, sizeof(int), cudaMemcpyDeviceToHost);

}
```
