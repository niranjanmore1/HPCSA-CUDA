```python
%%writefile sum.cu

#include <stdio.h>
#include <cuda.h>

__global__ void add(int *a, int *b, int *c)
{
    int i = threadIdx.x;
    c[i] = a[i] + b[i];
}

int main()
{
  int *a,*b,*c;

  cudaMallocManaged(&a,5*sizeof(int));
  cudaMallocManaged(&b,5*sizeof(int));
  cudaMallocManaged(&c,5*sizeof(int));

  for (int i = 0; i < 5; i++)
  {
    a[i] = i;
    b[i] = i * 10;
  }

  add<<<1,5>>>(a,b,c);

  cudaDeviceSynchronize();

  for (int i = 0; i < 5; i++)
    printf("%d\n", c[i]);

  cudaFree(a);
  cudaFree(b);
  cudaFree(c);

  return 0;
}
```

```python
!nvcc -arch=sm_70 sum.cu -o sum
```
