```python
%%writefile hello_world.cu
#include <stdio.h>
#include <cuda_runtime.h>

__global__ void kernel(void)
{
  printf ("hello from GPU\n");
}

void cpu_print(void)
{
  printf ("hello from CPU\n");
}



int main()
{
  kernel <<<1, 1>>>();
  kernel <<<1, 1>>>();
  kernel <<<1, 1>>>();
  cpu_print();
  cpu_print();
  cpu_print();
  cudaDeviceReset();
  return 0;
  }
```

```python
!nvcc hello_world.cu -o hello_world
```
